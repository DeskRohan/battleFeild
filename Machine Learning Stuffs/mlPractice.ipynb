```python
import seaborn as sns, matplotlib.pyplot as plt
from sklearn.datasets import fetch_california_housing

df = fetch_california_housing(as_frame=True).frame

for t,c in [('histplot','blue'),('boxplot','orange')]:
    plt.figure(figsize=(15,10))
    for i,col in enumerate(df,1):
        plt.subplot(3,3,i)

        if t=='histplot':
            sns.histplot(df[col], kde=True, bins=30, color=c)
        else:
            sns.boxplot(x=df[col], color=c)

        plt.title(col)

    plt.tight_layout()
    plt.show()
```

```python
import numpy as np, pandas as pd, matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.decomposition import PCA

i=load_iris()
d=PCA(2).fit_transform(i.data)
df=pd.DataFrame(d,columns=['PC1','PC2'])
df['L']=i.target

plt.figure(figsize=(8,6))
for j,c in enumerate(['r','g','b']):
    x=df[df.L==j]
    plt.scatter(x.PC1,x.PC2,label=i.target_names[j],color=c)

plt.title('PCA on Iris Dataset')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.legend(); plt.grid(); plt.show()
```

```python
import pandas as pd

def find_s_algorithm(file_path):
    data = pd.read_csv(file_path)

    print("Training data:")
    print(data)

    h = ['?']*(len(data.columns)-1)

    for _, row in data.iterrows():
        if row.iloc[-1] == 'Yes':
            for i, v in enumerate(row[:-1]):
                h[i] = v if h[i] in ['?', v] else '?'

    return h

file_path = "https://raw.githubusercontent.com/DeskRohan/mlCSV/main/training_data.csv"

hypothesis = find_s_algorithm(file_path)

print("\nThe final hypothesis is:", hypothesis)
```

```python
import numpy as np, matplotlib.pyplot as plt 
from collections import Counter 
 
d=np.random.rand(100) 
td,vd=d[:50],d[50:] 
tl=["Class1" if x<=.5 else "Class2" for x in td] 
ks=[1,2,3,4,5,20,30] 
 
f=lambda t,k:Counter([y for _,y in sorted((abs(t-td[i]),tl[i]) for i in range(50))[:k]]).most_common(1)[0][0] 
 
for k in ks: 
    r=[f(x,k) for x in vd] 
    print(f"\nResults for k={k}") 
    [print(f"x{i+51} ({vd[i]:.4f}) -> {r[i]}") for i in range(50)] 
 
    plt.figure(figsize=(10,6)) 
    plt.scatter(td,[0]*50,c=['blue' if x=="Class1" else 'red' for x in tl],label='Train') 
    plt.scatter([vd[i] for i in range(50) if r[i]=="Class1"],[1]*r.count("Class1"),c='blue',marker='x',label='Class1') 
    plt.scatter([vd[i] for i in range(50) if r[i]=="Class2"],[1]*r.count("Class2"),c='red',marker='x',label='Class2') 
    plt.title(f"k={k}") 
    plt.grid() 
    plt.legend() 
    plt.show() 
```

```python
import numpy as np, matplotlib.pyplot as plt 
 
g=lambda x,xi,t:np.exp(-np.sum((x-xi)**2)/(2*t**2)) 
 
def lwr(x,X,y,t): 
    w=np.diag([g(x,i,t) for i in X]) 
    return x@np.linalg.inv(X.T@w@X)@X.T@w@y 
 
np.random.seed(42) 
X=np.linspace(0,2*np.pi,100) 
y=np.sin(X)+.1*np.random.randn(100) 
Xb=np.c_[np.ones(X.shape),X] 
xt=np.linspace(0,2*np.pi,200) 
xtb=np.c_[np.ones(xt.shape),xt] 
t=.5 
yp=[lwr(i,Xb,y,t) for i in xtb] 
plt.figure(figsize=(10,6)) 
plt.scatter(X,y,c='red',label='Training Data',alpha=.7) 
plt.plot(xt,yp,c='blue',label=f'LWR Fit (tau={t})',lw=2) 
plt.xlabel('X'); plt.ylabel('y') 
plt.title('Loca ly Weighted Regression') 
plt.legend(); plt.grid(alpha=.3) 
plt.show() 
```

```python
import pandas as pd, matplotlib.pyplot as plt 
from sklearn.datasets import fetch_california_housing 
from sklearn.model_selection import train_test_split as tts 
from sklearn.linear_model import LinearRegression as LR 
from sklearn.pipeline import make_pipeline 
from sklearn.preprocessing import PolynomialFeatures as PF,StandardScaler as SS 
from sklearn.metrics import mean_squared_error as mse,r2_score as r2 
# Linear Regression 
d=fetch_california_housing(as_frame=True) 
X,y=d.data[["AveRooms"]],d.target 
x1,x2,y1,y2=tts(X,y,test_size=.2,random_state=42) 
m=LR().fit(x1,y1) 
p=m.predict(x2) 
plt.scatter(x2,y2,c='blue',label='Actual') 
plt.plot(x2,p,c='red',label='Predicted') 
plt.title("Linear Regression") 
plt.legend(); plt.show() 
print("Linear Regression") 
print("MSE:",mse(y2,p),"\nR2:",r2(y2,p)) 
# Polynomial Regression 
u="https://archive.ics.uci.edu/ml/machine-learning-databases/auto-mpg/auto-mpg.data" 
c=["mpg","cyl","disp","hp","wt","acc","yr","org"] 
d=pd.read_csv(u,sep='\s+',names=c,na_values="?").dropna() 
X,y=d[["disp"]],d.mpg 
x1,x2,y1,y2=tts(X,y,test_size=.2,random_state=42) 
m=make_pipeline(PF(2),SS(),LR()).fit(x1,y1) 
p=m.predict(x2) 
plt.scatter(x2,y2,c='blue',label='Actual') 
plt.scatter(x2,p,c='red',label='Predicted') 
plt.title("Polynomial Regression") 
plt.legend(); plt.show() 
print("\nPolynomial Regression") 
print("MSE:",mse(y2,p),"\nR2:",r2(y2,p)) 
```
